```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
img=cv2.imread('/content/drive/MyDrive/cv img/messi.jpg',cv2.IMREAD_GRAYSCALE)
```

```python
if img is None:
  print("error:image not found")
else:
  plt.figure(figsize=(12,10))
  plt.subplot(3,3,1)
  plt.imshow(img,cmap='gray')
  plt.title('original image')
  plt.axis('off')
```

```python
for i in range(8):
  bit_plane=(img>>i)&1
  vis_plane=bit_plane*255
  plt.subplot(3,3,i+2)
  plt.imshow(vis_plane,cmap='gray')
  plt.title(f'bit plane {i+1}')
  plt.axis('off')
plt.tight_layout()
plt.show()
```
