```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
def show_comparision(original,blur3,blur5,blur11):
  plt.figure(figsize=(12,10))
  plt.subplot(2,2,1)
  plt.imshow(original,cmap='gray')
  plt.title("original grayscale image")
  plt.axis('off')
  plt.subplot(2,2,2)
  plt.imshow(blur3,cmap='gray')
  plt.title("3x3 kernel")
  plt.axis('off')
  plt.subplot(2,2,3)
  plt.imshow(blur5,cmap='gray')
  plt.title("5x5 kernel")
  plt.axis('off')
  plt.subplot(2,2,4)
  plt.imshow(blur11,cmap='gray')
  plt.title("11x11 kernel")
  plt.axis('off')
  plt.show()
```

```python
img1=cv2.imread('/content/drive/MyDrive/cv img/images cv.jpg')
```

```python
if img1 is None:
  print("error")
else:
  kernel_3x3=np.ones((3,3),np.float32)/9
  kernel_5x5=np.ones((5,5),np.float32)/25
  kernel_11x11=np.ones((11,11),np.float32)/49
```

```python
gray_img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
blur3 = cv2.filter2D(gray_img1, -1, kernel_3x3)
blur5 = cv2.filter2D(gray_img1, -1, kernel_5x5)
blur11 = cv2.filter2D(gray_img1, -1, kernel_11x11)
show_comparision(gray_img1, blur3, blur5, blur11)
```
